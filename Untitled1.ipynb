```python
import os
import shutil
from PIL import Image

final_path = "/content/dataset_cabai/Chili Leaf Disease Original Dataset/Chili Leaf Disease Original Dataset"

print("Mulai pembersihan file rusak dan tersembunyi...")

for root, dirs, files in os.walk(final_path):
    # 1. Hapus folder hidden bawaan Colab
    if '.ipynb_checkpoints' in dirs:
        shutil.rmtree(os.path.join(root, '.ipynb_checkpoints'))
        print(f"Folder .ipynb_checkpoints di {root} dihapus.")

    # 2. Cek integritas setiap gambar dan hapus jika corrupt
    for file in files:
        file_path = os.path.join(root, file)
        try:
            img = Image.open(file_path)
            img.verify() # Validasi apakah file benar-benar gambar
        except Exception:
            print(f"Menghapus file invalid/rusak: {file}")
            os.remove(file_path)

print("\nPembersihan selesai! Silakan jalankan ulang sel training (model.fit) kamu.")
```

```python
import tensorflow as tf

# Tentukan path direktori, ukuran gambar, dan batch size
DATA_DIR = "/content/dataset_cabai/Chili Leaf Disease Original Dataset/Chili Leaf Disease Original Dataset"
IMG_SIZE = (224, 224) # Resolusi optimal untuk EfficientNetB0
BATCH_SIZE = 32

# 1. Memuat data untuk Training (80%)
train_dataset = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

# 2. Memuat data untuk Validation (20%)
val_dataset = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
import numpy as np

class_names = train_dataset.class_names
print("Daftar Kelas:", class_names)

# Fungsi untuk menghitung jumlah gambar per kelas
def hitung_per_kelas(dataset, nama_dataset):
    jumlah = {kelas: 0 for kelas in class_names}
    # Iterasi isi dataset
    for _, labels in dataset:
        for label in labels:
            jumlah[class_names[label.numpy()]] += 1

    print(f"\nJumlah pada {nama_dataset}:")
    for kelas, total in jumlah.items():
        print(f"- {kelas}: {total} gambar")

# Tampilkan hasilnya
hitung_per_kelas(train_dataset, "Data Training")
hitung_per_kelas(val_dataset, "Data Validation")
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Mengumpulkan semua label dari data training
train_labels = np.concatenate([y.numpy() for x, y in train_dataset], axis=0)

# Menghitung bobot kelas yang seimbang
weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_labels),
    y=train_labels
)

# Menyimpan dalam format dictionary yang dibutuhkan Keras
class_weights_dict = {i: weight for i, weight in enumerate(weights)}

print("Bobot per kelas (Class Weights):")
for i, weight in class_weights_dict.items():
    print(f"- Kelas {i} ({class_names[i]}): {weight:.4f}")
```

```python
AUTOTUNE = tf.data.AUTOTUNE

# Cache menyimpan data di memori setelah epoch pertama
# Shuffle mengacak data training agar model tidak menghafal urutan
# Prefetch menyiapkan data batch berikutnya saat GPU sedang bekerja
train_dataset = train_dataset.cache().shuffle(1000).prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.cache().prefetch(buffer_size=AUTOTUNE)

print("Pipeline data berhasil dioptimasi!")
```

```python
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers
import tensorflow as tf

# 1. Membuat layer Data Augmentation
data_augmentation = tf.keras.Sequential([
  layers.RandomFlip("horizontal_and_vertical"),
  layers.RandomRotation(0.2),
])

# 2. Memuat model dasar EfficientNetB0 (Pre-trained pada ImageNet)
base_model = EfficientNetB0(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Membekukan (freeze) model dasar agar bobotnya tidak rusak saat awal training
base_model.trainable = False

# 3. Merangkai Arsitektur Akhir
inputs = tf.keras.Input(shape=(224, 224, 3))
x = data_augmentation(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.2)(x) # Mencegah overfitting tambahan
outputs = layers.Dense(6, activation='softmax')(x) # 6 sesuai jumlah kelas

model = tf.keras.Model(inputs, outputs)

# 4. Kompilasi Model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Menampilkan ringkasan arsitektur
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Menyimpan model terbaik & menghentikan training jika tidak ada kemajuan
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
model_checkpoint = ModelCheckpoint('model_cabai_merah.h5', monitor='val_accuracy', save_best_only=True)

print("Memulai pelatihan model...")
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=20,
    class_weight=class_weights_dict, # Pastikan variabel class_weights_dict sudah ada di memori
    callbacks=[early_stopping, model_checkpoint]
)
```

```python
import matplotlib.pyplot as plt

# Mengambil data metrik dari history
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(1, len(acc) + 1)

# Mengatur kanvas grafik
plt.figure(figsize=(14, 5))

# 1. Grafik Akurasi (Accuracy)
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy', marker='o')
plt.plot(epochs_range, val_acc, label='Validation Accuracy', marker='o')
plt.legend(loc='lower right')
plt.title('Grafik Akurasi')
plt.grid(True)

# 2. Grafik Loss (Tingkat Kesalahan)
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss', marker='o')
plt.plot(epochs_range, val_loss, label='Validation Loss', marker='o')
plt.legend(loc='upper right')
plt.title('Grafik Loss')
plt.grid(True)

plt.show()
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

print("Sedang memproses prediksi pada data validasi...")
y_true = []
y_pred_probs = []

# Mengumpulkan prediksi dan label asli
for images, labels in val_dataset:
    y_true.extend(labels.numpy())
    preds = model.predict(images, verbose=0) # <-- Sudah diperbaiki menjadi 'model'
    y_pred_probs.extend(preds)

y_true = np.array(y_true)
y_pred = np.argmax(y_pred_probs, axis=1)

# Menampilkan metrik evaluasi
print("\n=== CLASSIFICATION REPORT ===")
print(classification_report(y_true, y_pred, target_names=class_names))

# Menggambar Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix')
plt.ylabel('Label Asli')
plt.xlabel('Prediksi Model')
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers, models, callbacks

# 1. Path direktori dataset gabungan
final_path = "/content/dataset_cabai/Chili Leaf Disease Original Dataset/Chili Leaf Disease Original Dataset"

# 2. Augmentasi Data (Krusial untuk adaptasi kondisi lahan)
# Kita menggunakan augmentasi pada data training, tapi tidak pada data validasi
train_datagen = ImageDataGenerator(
    validation_split=0.2,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.3, # Membantu mengabaikan mulsa/tanah
    horizontal_flip=True,
    fill_mode='nearest'
)

# Data validasi hanya displit, tanpa distorsi
val_datagen = ImageDataGenerator(validation_split=0.2)

# Menyiapkan aliran data
train_generator = train_datagen.flow_from_directory(
    final_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

val_generator = val_datagen.flow_from_directory(
    final_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)

# 3. Membangun Arsitektur EfficientNetB0
base_model = EfficientNetB0(input_shape=(224, 224, 3), include_top=False, weights='imagenet')

# Freeze base_model agar bobot ImageNet tidak rusak di tahap awal
base_model.trainable = False

# Layer klasifikasi khusus untuk 6 kelas
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.4), # Mencegah overfitting
    layers.Dense(6, activation='softmax')
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Callbacks untuk menghentikan training jika sudah optimal
early_stop = callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
reduce_lr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3)

# 5. Eksekusi Pelatihan
print("\nMemulai proses pelatihan Tahap 1...")
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=20,
    callbacks=[early_stop],
    class_weight=class_weights_dict  # <-- Tambahkan baris ini
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

# Explicitly define final_path to ensure it's available in this cell
final_path = "/content/dataset_cabai/Chili Leaf Disease Original Dataset/Chili Leaf Disease Original Dataset"

# Add a check for the existence of the directory
if not os.path.exists(final_path):
    print(f"ERROR: Directory '{final_path}' not found.")
    print("Please ensure that the initial data setup cells (where the dataset is downloaded and processed) have been run successfully.")
    raise FileNotFoundError(f"Missing data directory: {final_path}. Please run setup cells.")

# 1. Menyiapkan aliran data validasi murni (tanpa acak/shuffle)
val_datagen = ImageDataGenerator(validation_split=0.2)
val_gen_eval = val_datagen.flow_from_directory(
    final_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

# 2. Menjalankan prediksi menggunakan model hasil fine-tuning
print("\nMemproses prediksi data validasi...")
Y_pred = model.predict(val_gen_eval)
y_pred_classes = np.argmax(Y_pred, axis=1)
true_classes = val_gen_eval.classes

# 3. Menghitung dan menggambar Confusion Matrix
cm = confusion_matrix(true_classes, y_pred_classes)
class_names = list(val_gen_eval.class_indices.keys())

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)

fig, ax = plt.subplots(figsize=(10, 8))
disp.plot(cmap=plt.cm.Blues, ax=ax, xticks_rotation=45)
plt.title('Confusion Matrix - Hasil Akhir (Fine-Tuning)')
plt.tight_layout()
plt.show()

# 4. Menampilkan laporan klasifikasi teks (opsional, untuk melihat precision/recall)
from sklearn.metrics import classification_report
print("\nLaporan Klasifikasi:\n")
print(classification_report(true_classes, y_pred_classes, target_names=class_names))
```
